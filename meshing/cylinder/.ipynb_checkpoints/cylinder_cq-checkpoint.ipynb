```python
import cadquery as cq
from jupyter_cadquery import show
from jupyter_cadquery.replay import replay, enable_replay

enable_replay()
show_object = replay

# ---------------------------
# PARAMETERS
# ---------------------------
R = 50.0
H = 200.0
slab_thickness = 5.0

# ---------------------------
# BASE CYLINDER (single parent solid)
# ---------------------------
base_cyl = (
    cq.Workplane("front")
    .circle(R)
    .extrude(H)
)

# ---------------------------
# SLAB CUTTER (shared geometry)
# ---------------------------
slab = (
    cq.Workplane("front")
    .rect(2 * R, slab_thickness)
    .extrude(H)
)

# ---------------------------
# REMOVE SLAB FROM CYLINDER
# ---------------------------
cyl_minus_slab = base_cyl.cut(slab)

# ---------------------------
# HALF-SPACE CUTTERS
# ---------------------------
upper_half_space = (
    cq.Workplane("front")
    .rect(2 * R, R)
    .extrude(H)
    .translate((0,  R/2, 0))
)

lower_half_space = (
    cq.Workplane("front")
    .rect(2 * R, R)
    .extrude(H)
    .translate((0, -R/2, 0))
)

# ---------------------------
# TRUE PARTITIONED REGIONS
# ---------------------------
fluid_half_1 = cyl_minus_slab.intersect(upper_half_space)
fluid_half_2 = cyl_minus_slab.intersect(lower_half_space)

solid_slab = slab  # slab is already exact and shared

# ---------------------------
# REGION LABELS (logical names)
# ---------------------------
regions = {
    "fluid_half_1": fluid_half_1,
    "solid_slab": solid_slab,
    "fluid_half_2": fluid_half_2,
}

# ---------------------------
# VISUALIZATION
# ---------------------------
show_object(
    fluid_half_2
)
```

```python

assembly = cq.Assembly()

assembly.add(fluid_half_1, name="fluid_half_1")
assembly.add(solid_slab, name="solid_slab")
assembly.add(fluid_half_2, name="fluid_half_2")

assembly.save("partitioned_cylinder.step")
```
